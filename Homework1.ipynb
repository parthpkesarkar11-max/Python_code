```python
class Father:     #singel Inheritance
    def show1(self):
        print("Parent 1")
class son(Father):
    def show(self):
        super().show1()
        print("Child 2")
A1=son()
A1.show()
```

```python
class School:
    def show1(self):
        print("Study 1")
class Education:           #Multiple Inheritance
    def show2(self):
        print("Knowledge 2")
class Student(School,Education):
    def show(self):
        super().show1()
        super().show2()
        print("Growth 3")
B1=Student()
B1.show()
```

```python
class Grandfather:
    def show1(self):
        print("A")             #Multilevel INheritance
class Father(Grandfather):
    def show2(self):
        print("B")
class Son(Father):
    def show3(self):
        super().show1()
        print("C")
C1=Son()
C1.show3()
C1.show2()
```

```python
class Boss:
    def show1(self):
      print("1")                # hierarchical Inheritance
class Emp1(Boss):
    def show(self):
        print("1.1")
class Emp2(Boss):
    def show(self):
        super().show1()
        print("1.2")
D1=Emp2()
D1.show()
```

```python
class A:
    def show1(self):
        print("1")
class B(A):                    #Hybrid Inheritance
    def show2(self):
        super().show1()
        print("2")
class C(A):
    def show3(self):
        super().show1()
        print("3")
class D(B,C):
    def show4(self):
        super().show2()
        super().show3()
        print("4")
E1=D()
E1.show4()
        
```
